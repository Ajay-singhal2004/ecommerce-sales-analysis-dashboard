```python
# step 1:import required libraries

import pandas as pd
import numpy as np
```

```python
# step 2: load dataset

df = pd.read_excel('Ecommerce_Unclean_Project.xlsx')
```

```python
# step 3:ecplore the dataset
#view first 5 records

df.head()
```

```python
# view the dataset information

df.info()
```

```python
# statistical summary
df.describe()
```

```python
#dataset shape
df.shape
```

```python
# step 4:check missing values
df.isnull().sum()
```

```python
#step 5:check duplicate records 
df.duplicated().sum()
```

```python
# remove duplicate records

df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
# step 6: replace invalid value
#replace common invalid values with nan

df.replace(['N/A','NULL',''],np.nan,inplace=True)
```

```python
df
```

```python
#convert discount column into numeric
df['Discount']=pd.to_numeric(df['Discount'],errors='coerce')
```

```python
# step 7: remove extra spaces
df= df.apply(lambda x: x.str.strip() if x.dtype =='object' else x)
```

```python
df
```

```python
# step 8: standardize text format
#convert text into ptoper case.

df['Customer_Name']=df['Customer_Name'].str.title()

df['City']=df['City'].str.title()

df['State']=df['State'].str.title()
```

```python
# step 9:validate email addresses
#keep only valid email record.

df=df[df['Email'].str.contains('@',na=False)]
```

```python
df
```

```python
# step 10 :conver date columns

df['Order_Date']=pd.to_datetime(df['Order_Date'],errors='coerce',dayfirst=True) 
df['Delivery_Date']=pd.to_datetime(df['Delivery_Date'],errors='coerce',dayfirst=True) 
```

```python
# step 11:convert numeric colmns

cols=['Qty','Unit_Price']

for c in cols:
    df[c]=pd.to_numeric(df[c],errors='coerce')

#clean discount sepatately

df['Discount']=pd.to_numeric(df['Discount'],errors='coerce')
```

```python
df
```

```python
# step 12 remove invalid quantity
# remove records having nagative or zero quantity

df=df[df['Qty']>0]
```

```python
# step 13 :fill missing values

df['Discount']=df['Discount'].fillna(0)

df['Phone']=df['Phone'].fillna('unknown')

df['Delivery_Date']=df['Delivery_Date'].fillna(df['Order_Date'])
```

```python
df.isnull().sum()
```

```python
# step 14 : feature engineering 
# total sales

df['Sales']=df['Qty']*df['Unit_Price']
```

```python
df
```

```python
#net amount

df['Net_Amount']=df['Sales']-(df['Sales']*df['Discount']/100)
```

```python
df
```

```python
#profit(Example: 20%)

df['Profit']=df['Net_Amount']*0.20
```

```python
df
```

```python
# order month
df['Month']=df['Order_Date'].dt.month_name()
```

```python
df
```

```python
#order year
df['year']=df['Order_Date'].dt.year
```

```python
df
```

```python
# weekday
df['weekday']=df['Order_Date'].dt.day_name()
```

```python
df
```

```python
# step 15: final data  validation 

df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
#step 16 :export clean data 
df.to_csv('Clean_Ecommerce.csv',index=False)
```

```python
# export dataframe to mysql

df.to_sql(
    name='orders',
    con=engine,
    if_exists='replace',
    index=False
)
```
